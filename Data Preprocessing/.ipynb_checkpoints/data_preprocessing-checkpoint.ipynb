# Outliers 
An outlier is an observation unusually far from the rest of a distribution. It can be a data-entry error, a legitimate rare case, or an important event.

```python
# using iqr method 

Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3 - Q1 # covering the middle 50% observations

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Fare"] < lower) |
    (df["Fare"] > upper)
]

print("Lower fence:", lower)
print("Upper fence:", upper)
print("Flagged rows:", len(outliers))
```

## Categorical encoding

Most scikit-learn estimators expect numerical feature representations. Encoding converts categorical information into a form that a model can use.

This is useful for nominal categories where there is no natural numerical order.

In production, prefer OneHotEncoder inside your preprocessing pipeline. It supports unseen categories, consistent transformations, and integration with cross-validation.

```python
df['Sex']
```

```python
#one hot 
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")

encoded = encoder.fit_transform(df[["Sex"]])

encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(["Sex"]),
    index=df.index
)

print(encoded_df.head())
```

# Feature scaling: standardization, normalization and robust scaling
Scaling changes the numerical ranges of features. It does not automatically make a model better.

### 1. StandardScaler

$z = \frac{x-\mu}{\sigma}$

Example: $x=40,\ \mu=30,\ \sigma=10$

$z = \frac{40-30}{10}=1$

### 2. MinMaxScaler

$x' = \frac{x-x_{\min}}{x_{\max}-x_{\min}}$

Example: $x=50,\ x_{\min}=0,\ x_{\max}=100$

$x' = \frac{50-0}{100-0}=0.5$

### 3. RobustScaler

$x' = \frac{x-\text{Median}}{Q3-Q1}$

Example: $x=40,\ \text{Median}=30,\ IQR=10$

$x' = \frac{40-30}{10}=1$

```python
# standard scaler
# standardisation
scaler = StandardScaler()

values = np.array([[10], [20], [30]])

scaled = scaler.fit_transform(values)
print(scaled)
```

```python
# min max scaler
# normalization

scaler = MinMaxScaler()
scaled = scaler.fit_transform(values)
scaled
```

```python
# Robust Scaler
# used when outliers are present gets less effected by outliers 
scaler = RobustScaler()
scaled = scaler.fit_transform(values)
scaled
```
