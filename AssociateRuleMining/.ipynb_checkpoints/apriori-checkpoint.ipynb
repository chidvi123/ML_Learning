```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
transactions = [
    ["Bread", "Milk"],
    ["Bread", "Butter", "Milk"],
    ["Bread", "Butter"],
    ["Milk", "Butter"],
    ["Bread", "Milk", "Butter"]
]
```

```python
encoder = TransactionEncoder()

encoded_data = encoder.fit_transform(transactions)

df = pd.DataFrame(
    encoded_data,
    columns=encoder.columns_
)

df
```

```python
frequent_itemsets = apriori(
    df,
    min_support=0.4,
    use_colnames=True
)

frequent_itemsets
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.6
)

rules[[
    "antecedents",
    "consequents",
    "support",
    "confidence",
    "lift"
]].sort_values(
    by="lift",
    ascending=False
)
```

```python
for _, row in rules.iterrows():
    print(f"Rule: {set(row['antecedents'])} -> {set(row['consequents'])}")
    print(f"Support: {row['support']}")
    print(f"Confidence: {round(row['confidence'],2)}")
    print(f"Lift: {round(row['lift'],2)}")
    print("-" * 40)
```

# Market Basket Analysis and Apriori Algorithm

## 1. Market Basket Analysis

Market Basket Analysis is a data mining technique used to discover relationships
between items that frequently occur together in transactions.

It is commonly used in retail and e-commerce to identify products that customers
tend to purchase together.

### Example

A supermarket analyzes customer transactions and discovers that customers who
purchase bread often purchase milk as well.

This relationship can be represented as an association rule:

Bread → Milk

This means we are investigating how often milk is purchased in transactions
that contain bread. It does not mean bread causes customers to purchase milk.

### Applications

- Product recommendations
- Cross-selling and promotions
- Product placement
- Frequently purchased product combinations
- E-commerce recommendation systems

## 2. Association Rules

An association rule represents a relationship between itemsets.

A → B

- Antecedent (A): The item or itemset on the left side.
- Consequent (B): The item or itemset on the right side.

Example:

Bread → Milk

This rule examines how frequently milk appears in transactions containing bread.

The three important metrics used to evaluate association rules are:

1. Support
2. Confidence
3. Lift

## 3. Support

Support measures how frequently an item or combination of items appears across
all transactions.

### Formula

Support(A) = Number of transactions containing A / Total transactions

For an association rule A → B:

Support(A → B) = Number of transactions containing both A and B / Total transactions

Equivalently:

Support(A → B) = Support(A ∪ B)

### Example

Suppose there are 5 transactions, and bread appears in 4 of them.

Support(Bread) = 4 / 5 = 0.80 = 80%

If bread and milk appear together in 3 transactions:

Support(Bread, Milk) = 3 / 5 = 0.60 = 60%

Support uses all transactions as its denominator.

## 4. Confidence

Confidence measures how frequently B occurs in transactions that contain A.

### Formula

Confidence(A → B) = Support(A ∪ B) / Support(A)

Equivalently:

Confidence(A → B) =
Number of transactions containing both A and B /
Number of transactions containing A

### Example

- Bread appears in 4 transactions.
- Bread and milk appear together in 3 transactions.

Confidence(Bread → Milk) = 3 / 4 = 0.75 = 75%

Interpretation:

Among transactions containing bread, 75% also contain milk.

Confidence depends on the direction of the rule. In general,
Confidence(A → B) and Confidence(B → A) can differ.

## 5. Lift

Lift measures how the confidence of an association rule compares with the
overall support of its consequent.

It helps determine whether the association is stronger or weaker than expected
from the consequent's baseline frequency.

### Formula

Lift(A → B) = Confidence(A → B) / Support(B)

Equivalently:

Lift(A → B) = Support(A ∪ B) / (Support(A) × Support(B))

### Interpretation

- Lift > 1: Positive association.
- Lift = 1: No association beyond the baseline rate.
- Lift < 1: Negative association.

### Example

Suppose:

Confidence(Bread → Milk) = 75%
Support(Milk) = 80%

Lift(Bread → Milk) = 0.75 / 0.80 = 0.9375

The lift is below 1, so milk is less likely to occur among bread transactions
than its overall purchase rate would suggest.

Lift measures association, not causation.

## 6. Difference Between Support, Confidence, and Lift

| Metric | Meaning |
|---|---|
| Support | How frequently an item or combination occurs across all transactions |
| Confidence | Among transactions containing A, how frequently does B also occur? |
| Lift | How does the conditional occurrence rate compare with B's overall rate? |

Memory trick:

- Support = Frequency
- Confidence = Conditional frequency
- Lift = Association relative to the baseline

## 7. Apriori Algorithm

Apriori is an algorithm used to discover frequent itemsets in transaction data.

It is commonly used in Market Basket Analysis to find item combinations that
meet a minimum support threshold.

Apriori works by generating candidate itemsets, counting their support, and
pruning combinations that cannot be frequent.

### Important Terms

**Candidate itemsets (Ck)**

Potential itemsets whose support needs to be evaluated.

**Frequent itemsets (Lk)**

Itemsets that meet the minimum support threshold.

Here, k represents the number of items in an itemset.

- C1: Candidate 1-itemsets
- L1: Frequent 1-itemsets
- C2: Candidate 2-itemsets
- L2: Frequent 2-itemsets

## 8. Minimum Support

Minimum support is the threshold used to determine whether an itemset is
frequent enough to retain.

### Formula

Minimum support count = Minimum support × Total transactions

### Example

Suppose:

Total transactions = 5
Minimum support = 40%

Minimum support count = 0.40 × 5 = 2

Therefore, an itemset must occur in at least 2 transactions to qualify as frequent.

Minimum support can be specified as a proportion or percentage conceptually.
In mlxtend's apriori() function, min_support is supplied as a proportion.

For example:

min_support = 0.4

means 40% support.

### How to Choose Minimum Support

There is no universally correct minimum support threshold.

The choice depends on:

- Number of transactions
- Business requirements
- Whether common or rare patterns are important
- Number and usefulness of the resulting itemsets

In practice, experiment with multiple thresholds and inspect the resulting
frequent itemsets and association rules.

A lower threshold generally produces more frequent itemsets, while a higher
threshold generally produces fewer.

Minimum support filters itemsets by frequency. It does not directly establish
whether an association rule is useful.

## 9. Apriori Property

The Apriori property states:

If an itemset is frequent, all its non-empty subsets must also be frequent.

Equivalently:

If an itemset is infrequent, all its supersets must also be infrequent.

### Example

Suppose:

{Bread, Milk} is infrequent.

Then:

{Bread, Milk, Butter}

cannot be frequent.

There is no need to evaluate the larger itemset because one of its subsets
already fails the minimum support requirement.

This property allows Apriori to prune candidates and avoid unnecessary counting.

## 10. Complete Apriori Example

Consider the following supermarket transactions.

| Transaction | Items |
|---|---|
| T1 | Bread, Milk |
| T2 | Bread, Butter, Milk |
| T3 | Bread, Butter |
| T4 | Milk, Butter |
| T5 | Bread, Milk, Butter |

Set minimum support = 40%.

Minimum support count = 0.40 × 5 = 2.

An itemset must appear in at least 2 transactions.

### Step 1: Generate C1

Count the individual items.

| Candidate | Support count | Support |
|---|---:|---:|
| {Bread} | 4 | 80% |
| {Milk} | 4 | 80% |
| {Butter} | 4 | 80% |

All candidates pass the threshold.

L1 = {Bread}, {Milk}, {Butter}

### Step 2: Generate C2

Combine the frequent individual items into pairs.

C2:

- {Bread, Milk}
- {Bread, Butter}
- {Milk, Butter}

Count their occurrences.

| Candidate | Support count | Support |
|---|---:|---:|
| {Bread, Milk} | 3 | 60% |
| {Bread, Butter} | 3 | 60% |
| {Milk, Butter} | 3 | 60% |

All pairs pass the threshold.

L2 contains all three pairs.

### Step 3: Generate C3

Combine frequent pairs to consider the triple:

{Bread, Milk, Butter}

Before counting, check its 2-item subsets:

- {Bread, Milk}: frequent
- {Bread, Butter}: frequent
- {Milk, Butter}: frequent

All subsets are frequent, so the candidate survives pruning.

Now count the transactions containing all three items.

The triple appears in T2 and T5.

Support count = 2

Support = 2 / 5 = 40%

The triple passes the threshold.

L3 = {Bread, Milk, Butter}

### Step 4: Generate C4

No 4-itemset can be generated because the dataset contains only three distinct
items.

The algorithm stops.

### Final Frequent Itemsets

- L1: Three individual items
- L2: Three pairs
- L3: One triple

The sequence is:

C1 → L1 → C2 → L2 → C3 → L3 → Stop

C represents candidate itemsets.
L represents frequent itemsets.

### Step 5: Generate Association Rules

From {Bread, Milk}, we can generate:

- Bread → Milk
- Milk → Bread

For Bread → Milk:

Support = 3 / 5 = 60%

Confidence = 3 / 4 = 75%

Lift = 0.75 / 0.80 = 0.9375

The rule has 75% confidence but lift below 1.

This illustrates why high confidence alone does not guarantee a positive
association.

## 11. Apriori Workflow

1. Start with transaction data.
2. Set minimum support.
3. Generate candidate 1-itemsets (C1).
4. Count support and retain frequent itemsets (L1).
5. Generate candidate 2-itemsets (C2) from frequent itemsets.
6. Count support and retain frequent itemsets (L2).
7. Generate larger candidates and prune using the Apriori property.
8. Repeat until no more candidates can be generated.
9. Generate association rules from frequent itemsets.
10. Evaluate rules using support, confidence, and lift.

## 12. Python Implementation Using mlxtend

### What is mlxtend?

mlxtend stands for Machine Learning Extensions.

It is a Python library that provides additional machine learning and data
analysis utilities, including functions for frequent itemset mining and
association rule generation.

Functions used:

- TransactionEncoder: Converts transactions into a Boolean table.
- apriori(): Finds frequent itemsets.
- association_rules(): Generates association rules and their metrics.

### Step 1: Install the library

    %pip install mlxtend

### Step 2: Import libraries

    import pandas as pd
    from mlxtend.preprocessing import TransactionEncoder
    from mlxtend.frequent_patterns import apriori, association_rules

### Step 3: Create transaction data

    transactions = [
        ["Bread", "Milk"],
        ["Bread", "Butter", "Milk"],
        ["Bread", "Butter"],
        ["Milk", "Butter"],
        ["Bread", "Milk", "Butter"]
    ]

### Step 4: Encode transactions

Apriori expects a one-hot encoded table.

True means an item is present in a transaction.
False means it is absent.

    encoder = TransactionEncoder()

    encoded_data = encoder.fit_transform(transactions)

    df = pd.DataFrame(
        encoded_data,
        columns=encoder.columns_
    )

    df

fit_transform() learns the item vocabulary and encodes the supplied transactions.

For new transactions, use transform() on the already-fitted encoder to preserve
the same column mapping.

### Step 5: Find frequent itemsets

    frequent_itemsets = apriori(
        df,
        min_support=0.4,
        use_colnames=True
    )

    frequent_itemsets

min_support=0.4 means a minimum support proportion of 40%.

use_colnames=True displays item names rather than numeric column positions.

The function handles candidate generation and pruning internally. Its output
contains the frequent itemsets, not every intermediate candidate set.

### Step 6: Generate association rules

    rules = association_rules(
        frequent_itemsets,
        metric="confidence",
        min_threshold=0.6
    )

    rules[[
        "antecedents",
        "consequents",
        "support",
        "confidence",
        "lift"
    ]].sort_values(
        by="lift",
        ascending=False
    )

This generates rules with confidence of at least 0.6, or 60%.

The metric parameter specifies which rule metric is used for threshold filtering.

### Step 7: Filter useful rules

    useful_rules = rules[
        (rules["confidence"] >= 0.6) &
        (rules["lift"] > 1)
    ]

    useful_rules[[
        "antecedents",
        "consequents",
        "support",
        "confidence",
        "lift"
    ]]

This retains rules with at least 60% confidence and lift greater than 1.

An empty result is possible. That does not mean the code failed; it means no
generated rule meets both chosen conditions.

## 13. Minimum Support vs Minimum Confidence

| Parameter | Purpose |
|---|---|
| Minimum support | Filters frequent itemsets based on how frequently they occur |
| Minimum confidence | Filters association rules based on conditional occurrence frequency |
| Lift | Helps interpret the strength and direction of association relative to the baseline |

These thresholds are independent and should be selected based on the dataset
and the business objective.

## 14. Limitations of Apriori

- Candidate generation can produce many itemsets.
- Repeated scans of transaction data can be expensive.
- Performance can degrade on large datasets with many distinct items.
- Low minimum support can produce a very large number of patterns.
- High confidence does not necessarily imply a meaningful association.
- Association does not imply causation.

## 15. Apriori vs FP-Growth

| Apriori | FP-Growth |
|---|---|
| Generates candidate itemsets level by level | Builds a compressed FP-tree |
| Repeatedly counts candidate supports | Mines frequent patterns from the tree |
| Can generate many candidates | Avoids Apriori-style candidate generation |
| Can be expensive on large datasets | Often more efficient on suitable large datasets |

Both algorithms can find frequent itemsets for Market Basket Analysis.

## 16. Final Summary

Market Basket Analysis is the application used to discover relationships
between items in transaction data.

Apriori is one algorithm used for this task.

The algorithm generates candidate itemsets, counts support, prunes candidates
using the Apriori property, and retains frequent itemsets.

Association rules are then generated and evaluated using support, confidence,
and lift.

Key memory points:

- Support = Frequency across all transactions.
- Confidence = Conditional frequency of B given A.
- Lift = Association relative to B's baseline frequency.
- Ck = Candidate itemsets.
- Lk = Frequent itemsets.
- Minimum support = Frequency threshold.
- Apriori property = Every subset of a frequent itemset must be frequent.
- FP-Growth = Alternative frequent itemset mining algorithm that uses an FP-tree.
